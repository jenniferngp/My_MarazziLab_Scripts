```r
suppressMessages(
{
    library(data.table)
    library(rtracklayer)
    library(dplyr)
}) 
```

```r
# read gtf
gtf = "/dfs8/imarazzi_lab/share/project/pch-release-115/in/hg38.ensembl.transcriptome.annotation.gtf.gz"
df = data.frame(rtracklayer::import(gtf))
```

```r
table(df$seqnames)
```

```r
bed = df %>% 
    filter(type == "gene" & seqnames == 1) %>% 
    dplyr::select(seqnames, start, end, gene_id, score, strand)

unique(bed$seqnames)
length(unique(bed$gene_id))

fwrite(bed, "/data/homezvol1/jennipn6/personal/reference/public/ensembl/GRCh38_v115/hg38-ensembl-v115-chr1.genebody.bed", row.names = F, sep = "\t", col.names = F)

```
